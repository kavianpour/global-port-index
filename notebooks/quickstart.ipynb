Installs the package, builds a small port database from real UN/LOCODE data,
and runs some queries. No API keys needed. A full build (~116k rows) takes a
few minutes; this notebook builds without the optional WPI/OSM enrichment
step so it runs quickly.

```python
from port_index import build_database

meta = build_database("ports.sqlite")
meta["counts"]
```

```python
from port_index import PortIndex

idx = PortIndex("ports.sqlite")
print(idx.stats())
print(idx.resolve("NLRTM"))
print(idx.nearest(51.98, 3.60))
```

```python
for p in idx.search("singapore"):
    print(p.locode, p.name, p.country)
```
